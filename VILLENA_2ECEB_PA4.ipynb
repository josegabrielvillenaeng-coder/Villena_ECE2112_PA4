```python
import pandas as pd
import matplotlib.pyplot as plt
```

## **A. VISAYAS COMMUNICATION DATAFRAME**

```python
Boards = pd.read_excel('board2.xlsx')
```

```python
Boards['Average'] = Boards[['Math','Electronics','GEAS','Communication']].mean(axis=1)
```

```python
VisComm = Boards[(Boards['Hometown'] == 'Visayas') & (Boards['Track'] == 'Communication')][["Name","Gender","Math","Electronics","Average"]]
```

```python
print('Number of Rows:',len(VisComm))
VisComm
```

## **B. VISAYAS FEMALE DATAFRAME**

```python
df = Boards[(Boards['Hometown'] == 'Visayas') & (Boards['Gender'] == 'Female')][["Name","Track","GEAS","Electronics","Average"]]
VisFemale = df[ df['Average'] >= 60]
print('\nTable of VisFemale:\n')
display(df)
print('\nTable of VisFemale with an average score of atlease 60:\n')
display(VisFemale)
```

## **C. CATEGORY-AVERAGE VISUALIZATION**

```python
## C(a)

track_mean = Boards.groupby("Track")["Average"].mean()
gender_mean = Boards.groupby("Gender")["Average"].mean()
hometown_mean = Boards.groupby("Hometown")["Average"].mean()
```

```python
## C(b)

print("Mean Average by Track:")
print(track_mean)
print("\nMean Average by Gender:")
print(gender_mean)
print("\nMean Average by Hometown:")
print(hometown_mean)
```

```python
## C(c)

fig, axes = plt.subplots(1, 3, figsize=(15,6))

track_mean.plot(kind="bar", ax=axes[0])
axes[0].set_title("Average by Track")
axes[0].set_xlabel("Track")
axes[0].set_ylabel("Mean Average")

gender_mean.plot(kind="bar", ax=axes[1])
axes[1].set_title("Average by Gender")
axes[1].set_xlabel("Gender")
axes[1].set_ylabel("Mean Average")

hometown_mean.plot(kind="bar", ax=axes[2])
axes[2].set_title("Average by Hometown")
axes[2].set_xlabel("Hometown")
axes[2].set_ylabel("Mean Average")

plt.tight_layout()
plt.show()
```

```python
## C(d)

Htrack = track_mean.idxmax()
HGender = gender_mean.idxmax()
Hhometown = hometown_mean.idxmax()

print('1. The track with the highest mean:', Htrack)
print('2. The gender with the highest mean:', HGender)
print('3. The hometown with the highest mean:', Hhometown)
```
